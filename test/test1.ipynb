```python
import tushare as ts
import pandas as pd
import time
from DONOTGIT_API import request_API

token = request_API()
ts.set_token(token)
pro = ts.pro_api()

# tech_stocks = ["AAPL", "MSFT", "NVDA", "GOOGL", "AMZN", "META", "TSLA", "AMD", "INTC"]调取数据
tech_stocks = ["GOOG"]
all_data = []

for code in tech_stocks:
    print(f"正在下载 {code} ...")
    df = pro.us_daily(
        ts_code=code,
        start_date="20240101",
        end_date="20260501"
    )
    all_data.append(df)
    time.sleep(0.3)

result = pd.concat(all_data, ignore_index=True)

print(result.head())
result.to_csv("us_tech_daily.csv", index=False, encoding="utf-8-sig")
```
